```python
import json
import random
from pathlib import Path

# =========================
# Einstellungen
# =========================
filename = "Construction_a10_o107_m5_an57_ar12.json"  # JSON liegt neben dem Notebook
out_file = Path(filename).with_name(Path(filename).stem + "_paper.json")

random.seed(42)  # Reproduzierbarkeit (optional)

# =========================
# JSON laden
# =========================
with open(filename, "r", encoding="utf-8") as f:
    data = json.load(f)

# =========================
# 1) Verfügbare Baustellennummern aus Aufträgen einsammeln
# =========================
auftraege = data.get("Auftraege", [])
if not isinstance(auftraege, list):
    raise ValueError("'Auftraege' ist nicht als Liste hinterlegt.")

verwendete_baustellen = sorted({
    a.get("Baustellennummer")
    for a in auftraege
    if isinstance(a.get("Baustellennummer"), int)
})
if not verwendete_baustellen:
    raise ValueError("Keine gültigen 'Baustellennummer'n in den Aufträgen gefunden.")

# =========================
# 2) Arbeitern BaustellenKenntnisse (1–3) zuweisen
# =========================
arbeiter = data.get("Arbeiter", [])
if not isinstance(arbeiter, list):
    raise ValueError("'Arbeiter' ist nicht als Liste hinterlegt.")

for w in arbeiter:
    k = random.randint(1, 3)
    if k > len(verwendete_baustellen):
        k = len(verwendete_baustellen)
    w["BaustellenKenntnisse"] = random.sample(verwendete_baustellen, k)

# =========================
# Schreiben
# =========================
with open(out_file, "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=4)

print("✅ Fertig.")
print(f"   Arbeiter ergänzt: {len(arbeiter)} mit 'BaustellenKenntnisse'")
print(f"   Verfügbare Nummern aus Aufträgen: {len(verwendete_baustellen)} -> {verwendete_baustellen[:10]}{' ...' if len(verwendete_baustellen) > 10 else ''}")
print(f"   Datei gespeichert: {out_file}")
```
