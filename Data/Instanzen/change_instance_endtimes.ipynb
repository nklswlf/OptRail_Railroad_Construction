```python
import os
import json

# Verzeichnis, von dem aus das Notebook läuft
hauptverzeichnis = os.getcwd()  # Aktuelles Arbeitsverzeichnis (Instanzen-Ordner)

def aktualisiere_endezeit_in_datei(dateipfad):
    """Aktualisiert die Endezeit in einer einzelnen JSON-Datei."""
    try:
        print(f"Überprüfe Datei: {dateipfad}")  # Debugging-Ausgabe
        with open(dateipfad, 'r', encoding='utf-8') as file:
            daten = json.load(file)  # JSON-Datei einlesen
        
        # Prüfen, ob der Schlüssel "Ende" existiert und angepasst werden muss
        if "Ende" in daten:
            print(f"Gefundene Endezeit: {daten['Ende']}")  # Debugging-Ausgabe
            if daten["Ende"] == "2024-08-31T00:00:00":
                daten["Ende"] = "2024-08-31T23:59:59"  # Neue Endezeit setzen
                
                # Geänderte Daten speichern
                with open(dateipfad, 'w', encoding='utf-8') as file:
                    json.dump(daten, file, ensure_ascii=False, indent=4)
                
                print(f"Die Endezeit wurde aktualisiert: {dateipfad}")
            else:
                print(f"Keine Änderung erforderlich für Datei: {dateipfad}")
        else:
            print(f"Schlüssel 'Ende' nicht gefunden in Datei: {dateipfad}")
    except Exception as e:
        print(f"Fehler beim Bearbeiten der Datei {dateipfad}: {e}")

def aktualisiere_endezeit_rekursiv(verzeichnis):
    """Rekursiv durch alle Verzeichnisse und Dateien iterieren und JSON-Dateien bearbeiten."""
    print(f"Durchsuche Verzeichnis: {verzeichnis}")  # Debugging-Ausgabe
    for root, _, dateien in os.walk(verzeichnis):
        for dateiname in dateien:
            if dateiname.endswith('.json'):  # Nur JSON-Dateien bearbeiten
                dateipfad = os.path.join(root, dateiname)
                aktualisiere_endezeit_in_datei(dateipfad)

# Starte die Aktualisierung
aktualisiere_endezeit_rekursiv(hauptverzeichnis)
```
