```python
import json
import os
from collections import Counter

# 🔹 JSON-Datei laden
with open("data.json", "r", encoding="utf-8") as file:
    data = json.load(file)

# 🔹 Doppelte Anbaugerätetypen pro Bestellposition prüfen
duplicate_entries = {}

for bestellposition in data.get("Bestellpositionen", []):
    anbaugeraete_typen = bestellposition.get("AnbaugeraeteTypen", [])
    
    # Zählen, wie oft jeder Typ vorkommt
    type_count = Counter(anbaugeraete_typen)
    
    # Finde doppelte Einträge (mehr als 1x vorkommend)
    duplicates = {typ: count for typ, count in type_count.items() if count > 1}

    if duplicates:
        duplicate_entries[bestellposition["ID"]] = duplicates

# 🔹 Ergebnisse ausgeben
if duplicate_entries:
    print("❌ Doppelte Anbaugerätetypen gefunden:")
    for bestell_id, duplicates in duplicate_entries.items():
        print(f"  Bestellposition {bestell_id}: {duplicates}")
else:
    print("✅ Keine doppelten Anbaugerätetypen gefunden!")
```
