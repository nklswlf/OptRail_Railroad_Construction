```python
import json
import os

# 🔹 Liste der gesuchten JSON-Dateien
instances = [
    "Construction_a3_o80_m10_an10_ar9_reduced.json",
    "Construction_a5_o96_m10_an10_ar10_reduced.json",
    "Construction_a10_o107_m5_an57_ar12.json",
    "Construction_a10_o114_m6_an57_ar11.json",
    "Construction_a10_o128_m6_an51_ar13.json",
    "Construction_a10_o144_m6_an53_ar12.json",
    "Construction_a15_o170_m9_an80_ar18.json",
    "Construction_a20_o236_m12_an106_ar24.json",
    "Construction_a25_o306_m13_an127_ar31.json",
    "Construction_a30_o355_m18_an148_ar42.json",
    "Construction_a40_o476_m22_an215_ar51.json",
    "Construction_a50_o578_m28_an276_ar66.json"
]

# 🔹 Ordner durchsuchen
current_directory = os.getcwd()  # Arbeitsverzeichnis, wo das Skript läuft
found_files = {}  # Speichert den Pfad zu jeder Datei

for root, dirs, files in os.walk(current_directory):
    for filename in files:
        if filename in instances:
            found_files[filename] = os.path.join(root, filename)

# 🔹 Fehlende Anbaugeräte speichern
fehlende_gesamt = {}

# 🔹 Alle gefundenen Dateien verarbeiten
for filename, file_path in found_files.items():
    try:
        # 📌 JSON laden
        with open(file_path, "r", encoding="utf-8") as file:
            data = json.load(file)

        # 📌 Geforderte Anbaugeräte extrahieren
        geforderte_anbaugeraete = set()
        for bestellposition in data.get("Bestellpositionen", []):
            geforderte_anbaugeraete.update(bestellposition.get("AnbaugeraeteTypen", []))

        # 📌 Vorhandene Anbaugeräte extrahieren
        vorhandene_anbaugeraete = {anbaugeraet["Typ"] for anbaugeraet in data.get("Anbaugeraete", [])}

        # 📌 Fehlende Anbaugeräte berechnen
        fehlende_anbaugeraete = geforderte_anbaugeraete - vorhandene_anbaugeraete

        if fehlende_anbaugeraete:
            print(f"❌ {filename}: Fehlende Anbaugeräte: {fehlende_anbaugeraete}")
            fehlende_gesamt[filename] = fehlende_anbaugeraete
        else:
            print(f"✅ {filename}: Alle geforderten Anbaugeräte sind vorhanden!")

    except json.JSONDecodeError:
        print(f"❌ Fehler beim Parsen der JSON-Datei: {filename}")

# 🔹 Fehlende Anbaugeräte in eine JSON-kompatible Form bringen
fehlende_gesamt_json = {filename: list(geraete) for filename, geraete in fehlende_gesamt.items()}

# 🔹 JSON speichern
fehlende_datei = os.path.join(current_directory, "fehlende_anbaugeraete.json")
with open(fehlende_datei, "w", encoding="utf-8") as f:
    json.dump(fehlende_gesamt_json, f, indent=4)

print(f"📂 Fehlende Anbaugeräte gespeichert unter: {fehlende_datei}")
```
