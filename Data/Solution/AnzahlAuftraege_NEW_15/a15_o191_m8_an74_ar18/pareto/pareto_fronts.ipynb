```python
import pandas as pd
import matplotlib.pyplot as plt

# Liste der CSV-Dateien und deren Achsentitel
csv_files = [
    ("MachineTransportDistance_pareto_results.csv", "MachineTransportDistance"),
    ("WorkerWorkDistance_pareto_results.csv", "WorkerWorkDistance"),
    ("MachineUsage_pareto_results.csv", "MachineUsage"),
    ("WorkerUsage_pareto_results.csv", "WorkerUsage")
]

# Pareto-Kurven erstellen
for file_name, x_label in csv_files:
    try:
        # CSV-Datei einlesen
        df = pd.read_csv(file_name)

        # Pareto-Kurve plotten
        plt.figure(figsize=(8, 6))
        plt.plot(
            df[x_label],  # x-Achse aus der Datei
            df["Construction Fulfillment"],  # y-Achse
            marker='o',
            linestyle='-',
            label=f"Pareto Front: {x_label}"
        )

        # Beschriftungen und Titel setzen
        plt.title(f"Pareto Front: {x_label} vs Construction Fulfillment")
        plt.xlabel(x_label)
        plt.ylabel("Construction Fulfillment")
        plt.grid(True)
        plt.legend()
        plt.show()

    except FileNotFoundError:
        print(f"Datei {file_name} wurde nicht gefunden.")
    except KeyError as e:
        print(f"Spalte fehlt in der Datei {file_name}: {e}")
```
