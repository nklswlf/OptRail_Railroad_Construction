```python
# Pareto Trade-off Visualizer for Secondary Objectives
# ====================================================

import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display
import numpy as np

# ================================
# Step 1: Load CSV
# ================================

# Lade deine Datei – Pfad ggf. anpassen
df = pd.read_csv("RealLife_2024_7_1_2/TPSA/ParetoFront.csv")

# Zeige die ersten Zeilen
display(df.head())

# ================================
# Step 2: Define selectable objectives
# ================================

secondary_objectives = [
    "Driver Violation",
    "Commute Distance",
    "Transport Machines",
    "Transport Attachments",
    "Machines",
    "Workers",
    "Attachments"
]

# ================================
# Step 3: Create Dropdown Widgets
# ================================

x_dropdown = widgets.Dropdown(
    options=secondary_objectives,
    value="Machines",
    description="X-Axis (cost):",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="50%")
)

y_dropdown = widgets.Dropdown(
    options=secondary_objectives,
    value="Commute Distance",
    description="Y-Axis (benefit):",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="50%")
)

# ================================
# Step 4: Trade-off Calculation
# ================================

def calculate_tradeoff(x_obj, y_obj):
    # Daten vorbereiten (sortiert nach X)
    sorted_df = df[[x_obj, y_obj]].sort_values(x_obj).dropna()
    x = sorted_df[x_obj].values
    y = sorted_df[y_obj].values

    # Differenzen berechnen
    dx = np.diff(x)
    dy = np.diff(y)

    with np.errstate(divide="ignore", invalid="ignore"):
        tradeoffs = dy / dx

    # Durchschnittlicher Trade-off
    valid = np.isfinite(tradeoffs)
    avg_tradeoff = np.mean(np.abs(tradeoffs[valid]))

    print(f"\n🔁 Average absolute trade-off between '{y_obj}' and '{x_obj}': {avg_tradeoff:.2f} Δ{y_obj} / Δ{x_obj}")
    return tradeoffs, x[:-1], y[:-1]

# ================================
# Step 5: Plotting Function
# ================================

def plot_tradeoff_with_metric(x_obj, y_obj):
    plt.figure(figsize=(8, 6))
    plt.scatter(df[x_obj], df[y_obj], alpha=0.8, color="tab:blue")
    plt.xlabel(x_obj)
    plt.ylabel(y_obj)
    plt.title(f"Trade-off: {y_obj} vs. {x_obj}")
    plt.grid(True)
    plt.tight_layout()
    plt.show()
    
    # Berechne und zeige Tradeoff
    calculate_tradeoff(x_obj, y_obj)

# ================================
# Step 6: Interactive Display
# ================================

widgets.interact(plot_tradeoff_with_metric, x_obj=x_dropdown, y_obj=y_dropdown)
```
