```python
def parse_solution_file(file_path):
    routes = {
        'x': {},  # machine_id: [from-to sequence]
        'y': {},  # worker_id: [from-to sequence]
        'z': {},  # attachment_id: [from-to sequence]
    }

    with open(file_path, 'r') as file:
        for line in file:
            if line.startswith('x[') and ' 1' in line:
                key = line.split('[')[1].split(']')[0]
                machine, f, t = map(str.strip, key.split(','))
                routes['x'].setdefault(machine, []).append((f, t))
            elif line.startswith('y[') and ' 1' in line:
                key = line.split('[')[1].split(']')[0]
                worker, f, t = map(str.strip, key.split(','))
                routes['y'].setdefault(worker, []).append((f, t))
            elif line.startswith('z[') and ' 1' in line:
                key = line.split('[')[1].split(']')[0]
                attach, f, t = map(str.strip, key.split(','))
                routes['z'].setdefault(attach, []).append((f, t))

    def build_path(transitions):
        path = []
        if not transitions:
            return path
        current = '0'
        visited = set()
        while True:
            for (f, t) in transitions:
                if f == current and (f, t) not in visited:
                    path.append(t)
                    visited.add((f, t))
                    current = t
                    break
            else:
                break
        return path

    for k in ['x', 'y', 'z']:
        for id_, transitions in routes[k].items():
            routes[k][id_] = ['0'] + build_path(transitions)

    return routes


# Beispielaufruf
solution_path = "solution_a20_o236_m12_an106_ar24.sol"
routes = parse_solution_file(solution_path)

# Ausgabe
print("Maschinenrouten (x):")
print(routes['x'])
print("\nArbeiterrouten (y):")
print(routes['y'])
print("\nAnbaugeräte-Routen (z):")
print(routes['z'])
```
