```python
import pandas as pd
from pathlib import Path

# Liste der Ordner
folders = ["Solutions_1", "Solutions_2", "Solutions_3"]
filename = "Metrics.csv"

# Liste der Metrikspalten, über die der Durchschnitt berechnet werden soll
metric_columns = [
    "Improvement_Time", "DCI", "Distribution Metric (DM)",
    "Hypevolume (Monte Carlo)", "PCI", "PF-Share", "Spacing", "Spread"
]

# Für jede Lösung separat die Durchschnittswerte berechnen und anzeigen
for folder in folders:
    path = Path(folder) / filename
    print(f"\n📁 Durchschnittswerte aus {folder}\n{'=' * (30 + len(folder))}")
    
    if path.exists():
        df = pd.read_csv(path)

        # Gruppiere nach Algorithmus und berechne Mittelwerte
        group_avg = df.groupby("Algorithm")[metric_columns].mean(numeric_only=True)

        display(group_avg)
    else:
        print(f"⚠️ Datei {path} nicht gefunden.")
```
